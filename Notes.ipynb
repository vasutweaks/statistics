$$W_x(s, \tau) = \frac{1}{\sqrt{s}} \int_{-\infty}^{\infty} x(t) \psi^* \left( \frac{t - \tau}{s} \right) dt$$

$$\psi_0(t) = \pi^{-1/4} e^{i \omega_0 t} e^{-t^2 / 2}$$

Where $\omega_0$ is the non-dimensional frequency. Usually, $\omega_0 = 6$ is chosen because at this value, the wavelet scale $s$ is almost exactly inversely proportional to the Fourier frequency ($f \approx \frac{1}{s}$), making the interpretation of periods highly intuitive.
