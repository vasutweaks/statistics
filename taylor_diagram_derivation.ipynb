# Derivation of Centered Pattern RMS Difference ($E'^2$)
This notebook details the derivation of the centered pattern root-mean-square (RMS) difference equation based on the definitions provided in Karl E. Taylor's 2001 paper.


### Step 1: State the definitions
First, we start with the definition of the centered pattern RMS difference, $E'$:
$${E'}^2 = \frac{1}{N}\sum_{n=1}^{N}[(f_n - \overline{f}) - (r_n - \overline{r})]^2$$

We also need the definition of the correlation coefficient, $R$:
$$R = \frac{\frac{1}{N}\sum_{n=1}^{N}(f_n - \overline{f})(r_n - \overline{r})}{\sigma_f\sigma_r}$$

By multiplying both sides by $\sigma_f\sigma_r$, we isolate the covariance term:
$$\frac{1}{N}\sum_{n=1}^{N}(f_n - \overline{f})(r_n - \overline{r}) = R\sigma_f\sigma_r$$

Recall the standard formulas for the variances:
$$\sigma_f^2 = \frac{1}{N}\sum_{n=1}^{N}(f_n - \overline{f})^2$$
$$\sigma_r^2 = \frac{1}{N}\sum_{n=1}^{N}(r_n - \overline{r})^2$$

### Step 2: Expand the ${E'}^2$ equation
Take the definition of ${E'}^2$ and expand the squared term inside the summation:
$${E'}^2 = \frac{1}{N}\sum_{n=1}^{N} \left[ (f_n - \overline{f})^2 - 2(f_n - \overline{f})(r_n - \overline{r}) + (r_n - \overline{r})^2 \right]$$

### Step 3: Distribute the summation
Separate the expanded expression into three distinct terms:
$${E'}^2 = \left[ \frac{1}{N}\sum_{n=1}^{N}(f_n - \overline{f})^2 \right] + \left[ \frac{1}{N}\sum_{n=1}^{N}(r_n - \overline{r})^2 \right] - 2 \left[ \frac{1}{N}\sum_{n=1}^{N}(f_n - \overline{f})(r_n - \overline{r}) \right]$$

### Step 4: Substitute the statistical definitions
Substitute the definitions of variance and covariance established in Step 1:
* The first term is the variance of the test field: $\sigma_f^2$
* The second term is the variance of the reference field: $\sigma_r^2$
* The third term is the covariance: $R\sigma_f\sigma_r$

Substituting these yields the final relationship:
$${E'}^2 = \sigma_f^2 + \sigma_r^2 - 2\sigma_f\sigma_r R$$

The full RMS difference is resolved into these two distinct components (overall bias $\bar{E}$ and pattern difference $E'$) which add together quadratically: $E^2 = \bar{E}^2 + {E'}^2$
